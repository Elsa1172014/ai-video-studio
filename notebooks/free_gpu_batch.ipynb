# Story Studio — free GPU batch runner
Makes the whole video from a plan exported by Story Studio (`story-studio.plan.json`). **No server, no paid credit.**

**Read first:** free notebook GPUs (Kaggle P100 / T4, Colab T4) are old and LTX-Video is built for bfloat16. It may run very slowly or fail on them, and free quotas change. This notebook has **not** been verified on those GPUs. If the run fails with a bfloat16 or out-of-memory error, use a GPU of your own (RTX 3060 or better) with the same command: `python batch.py plan.json --out out`.

Steps: enable a GPU and Internet in the notebook settings, upload `story-studio.plan.json` next to this notebook, then run the cells in order.
